**How would handle the Exceptions?**

We use `try:` and `except:` blocks to handle the exceptions in python.

* `try:` block: The code which you think is going to make an error is written in this block.
* `except:` block: If the code in the try: block made mistake then what to do is wriiten in this block.

```python
# Let's create a file
with open('sample.txt', 'w') as f:
  f.write('hello world')
```

```python
# try except demo
try:
  with open ('sample1.txt', 'r') as f:
    print(f.read())
except:
  print('File not found')
```

### else block:

**`else`:** The code which you want to excecute after the success of `try:`, you write it into the `else:` block.

```python
# except
try:
  f= open('sample1.txt', 'r') # giving wrong file

except FileNotFoundError:
  print('File sapadli nahi bhau')

except Exception as e:
  print(e)

else:
  print(f.read())
```

**That means there are two ways from the `try:` one is `except:` if there is an error in the code of try block and another is `else:` if try runs successfully.**.

## Custom Exception:

It is used for full control

```python
class SecurityError(Exception):
  def __init__(self, message):
    print(message)

  def logout(self):
    print('You are logged out from all devices')

class Google():

  def __init__(self, name, email, password, device):
    self.name= name
    self.email= email
    self.password= password
    self.device= device


  def login(self, email, password, device):
    if device != self.device:
      raise SecurityError('Someone is trying to login from different device')
    if email== self.email and password== self.password:
      print('Welcome')
    else:
      print('Wrong emial or password')

```

```python
obj= Google('keshav', 'keshav@gmail.com', '1234', 'android')

try:
  obj.login('keshav@gmail.com', '1234', 'android')

except SecurityError as e:
  e.logout()

else:
  print(obj.name)

finally:
  print('login/logout operration completed')
```
